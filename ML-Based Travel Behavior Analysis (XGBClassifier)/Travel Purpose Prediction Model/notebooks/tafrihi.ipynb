```python
import pandas as pd
import numpy as np 
```

```python
train_data = pd.read_csv('../data/train_data.csv')
train_data.isna().sum()
```

```python
test_data = pd.read_csv('../data/test_data.csv')
test_data
```

```python
# Filling missing value
mod = train_data.groupby('Vehicle')['VehicleType'].transform(lambda x : x.mode()[0])
train_data['VehicleType'] = train_data.groupby('Vehicle')['VehicleType'].fillna(mod)
mod = train_data.groupby('VehicleType')['VehicleClass'].transform(lambda x: x.mode()[0] if not x.mode().empty else 0)
train_data['VehicleClass'] = train_data.groupby('VehicleType')['VehicleClass'].fillna(mod)
mod = test_data.groupby('Vehicle')['VehicleType'].transform(lambda x : x.mode()[0])
test_data['VehicleType'] = test_data.groupby('Vehicle')['VehicleType'].fillna(mod)
mod = test_data.groupby('VehicleType')['VehicleClass'].transform(lambda x: x.mode()[0] if not x.mode().empty else 0)
test_data['VehicleClass'] = test_data.groupby('VehicleType')['VehicleClass'].fillna(mod)
```

```python
# Feature Creation
train_data['DepartureTime_month'] = train_data['DepartureTime'].apply(lambda x: pd.to_datetime(x).month)
test_data['DepartureTime_month'] = test_data['DepartureTime'].apply(lambda x: pd.to_datetime(x).month)

train_data['DepartureTime_hour'] = train_data['DepartureTime'].apply(lambda x: pd.to_datetime(x).hour)
test_data['DepartureTime_hour'] = test_data['DepartureTime'].apply(lambda x: pd.to_datetime(x).hour)

train_data['Created_month'] = train_data['Created'].apply(lambda x: pd.to_datetime(x).month)
test_data['Created_month'] = test_data['Created'].apply(lambda x: pd.to_datetime(x).month)

train_data['Created_hour'] = train_data['Created'].apply(lambda x: pd.to_datetime(x).hour)
test_data['Created_hour'] = test_data['Created'].apply(lambda x: pd.to_datetime(x).hour)

```

```python
# Repalcing True False Values
train_data['Male'] = train_data['Male'].replace({True: 1, False: 0})
test_data['Male'] = test_data['Male'].replace({True: 1, False: 0})
train_data['VehicleClass'] = train_data['VehicleClass'].replace({True: 1, False: 0 ,np.nan:-1})
test_data['VehicleClass'] = test_data['VehicleClass'].replace({True: 1, False: 0 , np.nan:-1})
train_data['TripReason'] = train_data['TripReason'].replace({'Work': 1, 'Int': 0 })
```

```python
# Scaling
from sklearn.preprocessing import MinMaxScaler
scaler_coupon = MinMaxScaler()
scaler_price = MinMaxScaler()

train_data['CouponDiscount'] = scaler_coupon.fit_transform(train_data[['CouponDiscount']])
train_data['Price'] = scaler_price.fit_transform(train_data[['Price']])

test_data['CouponDiscount'] = scaler_coupon.transform(test_data[['CouponDiscount']])
test_data['Price'] = scaler_price.transform(test_data[['Price']])
```

```python
# Encoding
dummy = pd.get_dummies(train_data['Vehicle'], prefix = 'Vehicle')
train_data = pd.concat([train_data, dummy], axis=1)
train_data = train_data.drop('Vehicle', axis=1)

dummy = pd.get_dummies(test_data['Vehicle'], prefix = 'Vehicle')
test_data = pd.concat([test_data, dummy], axis=1)
test_data = test_data.drop('Vehicle', axis=1)
```

```python
# Dropping Useless Data
train_data.drop(columns = ['CancelTime','UserID','HashPassportNumber_p','HashEmail','BuyerMobile','NationalCode','DepartureTime','Created','BillID','TicketID','From', 'To','VehicleType'],inplace = True)
test_data.drop(columns = ['CancelTime','UserID','HashPassportNumber_p','HashEmail','BuyerMobile','NationalCode','DepartureTime','Created','BillID','TicketID','From', 'To','VehicleType'],inplace = True)
train_data               
```

```python
# Data Selection
from sklearn.model_selection import train_test_split

x = train_data.drop(columns = ['TripReason']) 
y = train_data['TripReason']
x_train , x_test , y_train ,y_test  = train_test_split(x , y, test_size = 0.2,random_state = 535, stratify = y)
```

```python
# modeling 

import xgboost as xgb

model = xgb.XGBClassifier(max_depth = 7 ,learning_rate = 0.2 , n_estimators = 200 , subsample = 0.8 , min_child_weight = 1, gamma = 1 )
model.fit(x_train,y_train)

```

```python
# evaluate model
from sklearn.metrics import  f1_score, classification_report, confusion_matrix
y_pred = model.predict(x_test)
print(f1_score(y_test,y_pred))
```

```python
# predict test samples
submission = model.predict(test_data)
submission = pd.DataFrame(submission)
submission.columns = ['TripReason']
submission['TripReason'].replace({1 : 'Work' , 0 : 'Int'} , inplace = True)
submission
```
