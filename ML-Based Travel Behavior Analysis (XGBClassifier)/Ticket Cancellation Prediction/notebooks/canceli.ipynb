```python
import pandas as pd
import numpy as np
```

```python
train_data = pd.read_csv('../data/train_data.csv')
train_data
```

```python
test_data = pd.read_csv('../data/test_data.csv')
test_data
```

```python
# do some preprocessing 
```

```python
# Drop useless data
train_data.drop(columns = ['CancelTime','UserID','HashPassportNumber_p','HashEmail','BuyerMobile','NationalCode'],inplace = True)
test_data.drop(columns = ['UserID','HashPassportNumber_p','HashEmail','BuyerMobile','NationalCode'],inplace = True)
```

```python
# Fill missing values
mod = train_data.groupby('Vehicle')['VehicleType'].transform(lambda x : x.mode()[0])
train_data['VehicleType'] = train_data.groupby('Vehicle')['VehicleType'].fillna(mod)
mod = train_data.groupby('VehicleType')['VehicleClass'].transform(lambda x: x.mode()[0] if not x.mode().empty else 0)
train_data['VehicleClass'] = train_data.groupby('VehicleType')['VehicleClass'].fillna(mod)
mod = test_data.groupby('Vehicle')['VehicleType'].transform(lambda x : x.mode()[0])
test_data['VehicleType'] = test_data.groupby('Vehicle')['VehicleType'].fillna(mod)
mod = test_data.groupby('VehicleType')['VehicleClass'].transform(lambda x: x.mode()[0] if not x.mode().empty else 0)
test_data['VehicleClass'] = test_data.groupby('VehicleType')['VehicleClass'].fillna(mod)
```

```python
train_data['family'] = train_data.groupby('BillID')['TicketID'].transform('count')
test_data['family'] = test_data.groupby('BillID')['TicketID'].transform('count')
train_data['is with family'] = (train_data.groupby('BillID')['family'].transform(lambda x: (x > 1).astype(int)))
test_data['is with family'] = (test_data.groupby('BillID')['family'].transform(lambda x: (x > 1).astype(int)))
```

```python
train_data['blit_gap'] = (pd.to_datetime(train_data['DepartureTime'])- pd.to_datetime(train_data['Created'])).dt.days  
test_data['blit_gap'] = (pd.to_datetime(test_data['DepartureTime'])- pd.to_datetime(test_data['Created'])).dt.days
```

```python
train_data['has discount'] = train_data['CouponDiscount'].apply(lambda x :(x > 1))
test_data['has discount'] = test_data['CouponDiscount'].apply(lambda x :(x > 1))
train_data['has discount'] = train_data['has discount'].replace({True: 1, False: 0})
test_data['has discount'] = test_data['has discount'].replace({True: 1, False: 0})
```

```python
train_data['net_price'] = train_data['Price'] - train_data['CouponDiscount'] 
test_data['net_price'] = test_data['Price'] - test_data['CouponDiscount']
```

```python
train_data['Male'] = train_data['Male'].replace({True: 1, False: 0})
test_data['Male'] = test_data['Male'].replace({True: 1, False: 0})
train_data['TripReason'] = train_data['TripReason'].replace({'Work': 1, 'Int': 0})
test_data['TripReason'] = test_data['TripReason'].replace({'Work': 1, 'Int': 0})
```

```python
from sklearn.preprocessing import MinMaxScaler
Scaler = MinMaxScaler()

train_data['CouponDiscount'] = Scaler.fit_transform(train_data[['CouponDiscount']])
train_data['Price'] = Scaler.fit_transform(train_data[['Price']])
train_data['net_price'] = Scaler.fit_transform(train_data[['net_price']])
test_data['CouponDiscount'] = Scaler.fit_transform(test_data[['CouponDiscount']])
test_data['Price'] = Scaler.fit_transform(test_data[['Price']])
test_data['net_price'] = Scaler.fit_transform(test_data[['net_price']])
```

```python
from sklearn.preprocessing import LabelEncoder
Encoder = LabelEncoder()

Labels = ['From','To','Vehicle']
for i in Labels:
    train_data[i] = Encoder.fit_transform(train_data[i])
    test_data[i] = Encoder.fit_transform(test_data[i])
```

```python
train_data.drop(columns = ['Created','DepartureTime','BillID','TicketID','CouponDiscount','VehicleType'],inplace = True)
test_data.drop(columns = ['Created','DepartureTime','BillID','TicketID','CouponDiscount','VehicleType'],inplace = True)
train_data['VehicleClass'] = train_data['VehicleClass'].astype(float)
test_data['VehicleClass'] = test_data['VehicleClass'].astype(float)
```

```python
train_data
```

```python
# Data Selection
from sklearn.model_selection import train_test_split

x = train_data.drop(columns = ['Cancel']) 
y = train_data['Cancel']
x_train , x_test , y_train ,y_test  = train_test_split(x , y)
x_train
```

```python
# modeling
import xgboost as xgb
model = xgb.XGBClassifier(max_depth = 20 ,learning_rate =0.1,n_estimators=1000, min_child_weight = 1, gamma = 1)
model.fit(x_train,y_train)
```

```python
# evaluate model
from sklearn.metrics import  f1_score, classification_report, confusion_matrix
y_pred = model.predict(x_test)
print(f1_score(y_test,y_pred))
```

```python
# predict test samples
submission = model.predict(test_data)
submission = pd.DataFrame(submission)
submission.columns = ['Cancel']
submission
```
