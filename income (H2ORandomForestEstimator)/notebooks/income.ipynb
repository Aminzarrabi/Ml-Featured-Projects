```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
import h2o
from h2o.estimators import H2ORandomForestEstimator
```

```python
train_data = pd.read_csv('../data/train.csv')
train_data
```

```python
test_data = pd.read_csv('../data/test.csv')
test_data
```

```python
x = train_data.drop(columns=['income'])
y = train_data['income']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=313)
```

```python
h2o.init()

x_train_h2o = h2o.H2OFrame(pd.concat([x_train, y_train], axis = 1))
x_test_h2o = h2o.H2OFrame(x_test)
traget = 'income'

model = H2ORandomForestEstimator(ntrees = 100 , sample_rate = 0.7)
model.train(x = list(x.columns), y = traget, training_frame = x_train_h2o)


```

```python
y_pred_h2o = model.predict(x_test_h2o)
y_pred = h2o.as_list(y_pred_h2o['predict'], use_pandas=True).values.flatten()

y_test_numeric = y_test.replace({'<=50K': 0, '>50K': 1}).values
y_pred_numeric = pd.Series(y_pred).replace({'<=50K': 0, '>50K': 1}).values

print(f1_score(y_test_numeric, y_pred_numeric))
```

```python
test_data_h2o = h2o.H2OFrame(test_data)
y_pred = model.predict(test_data_h2o)
submission = h2o.as_list(y_pred['predict'], use_pandas=True).values
submission
```

```python
submission = pd.DataFrame(submission)
submission.columns =['income']
submission
```
